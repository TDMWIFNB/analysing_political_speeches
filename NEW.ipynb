```python
# Load necessary libraries for text preprocessing
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords

# Download stopwords and set Dutch stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('dutch'))  # Using Dutch stopwords since the dataset is in Dutch

# Step 1: Text Preprocessing
# Function to clean text
def clean_text(text):
    # Ensure the text is a string
    text = str(text)
    # Convert to lowercase
    text = text.lower()
    # Remove special characters, numbers, and keep only letters and spaces
    text = re.sub(r'[^a-z\s]', '', text)
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    # Remove Dutch stopwords
    text = ' '.join(word for word in text.split() if word not in stop_words)
    return text

# Load full data
df_full = pd.read_csv("all_speeches_final_filtered_with_features.csv")

# Clean the speech text
df_full['speech_text'] = df_full['speech_text'].apply(clean_text)

# Check a few examples to confirm cleaning worked
print("Sample of cleaned text (first 3 rows):")
print(df_full['speech_text'].head(3))

# Segment by speaker and party (already in dataset, confirm it’s ready for analysis)
print("\nNumber of unique speakers:", df_full['speaker_name'].nunique())
print("Number of unique parties:", df_full['speaker_party'].nunique())
print("Sample of speakers and parties (first 5 rows):")
print(df_full[['speaker_name', 'speaker_party']].head(5))

# Sampling 3000 data to train the topic model
df_sample = df_full.sample(n=3000, random_state=42).reset_index(drop=True)
texts_sample = df_sample["speech_text"].tolist()
```
