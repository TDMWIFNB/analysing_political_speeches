```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import os
```

```python
# Initial URL for Stemming data from OData API
url = 'https://gegevensmagazijn.tweedekamer.nl/OData/v4/2.0/Stemming?$top=250'
all_stemming = []

try:
    while url:
        # Make GET request to API
        response = requests.get(url)
        response.raise_for_status()  # Raise an error for bad responses
        
        # Parse JSON response
        data = response.json()
        stemming_records = data.get("value", [])  # Extract records from 'value' key
        
        # Add current page of records to total list
        all_stemming.extend(stemming_records)
        
        # Get next page URL if it exists
        url = data.get("@odata.nextLink", None)
        
        # Print progress
        print(f"Retrieved {len(stemming_records)} stemming records. Total so far: {len(all_stemming)}")
    
    if not all_stemming:
        print("No records retrieved from the API.")
    else:
        # Create DataFrame from collected data
        df = pd.DataFrame(all_stemming)
        
        # Define the output directory and file path
        output_dir = "data"
        output_file = f"{output_dir}/stemming.csv"
        
        # Create directory if it doesn’t exist
        os.makedirs(output_dir, exist_ok=True)
        
        # Save to CSV file
        df.to_csv(output_file, index=False)
        print(f"Successfully saved {len(all_stemming)} stemming records to {output_file}")

except requests.exceptions.RequestException as e:
    print(f"Error fetching data: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
